### One-way ANOVA
Let say we have k groups with n observations. So, we have N= nk observation in total. Let denote overall mean as $\bar{Y}$

- Total sum of squares $SST = \sum_{i=1}^{N} (Y_i - \bar{Y})^2$

- Sum of squares treatment $SSA = n (\bar{Y}_i - \bar{Y})^2$

- Sum of squares error $SSE = SST - SSA $

- Mean squares treatment $MSA = \frac{SSA}{k-1}$ where k is number of groups

- Mean squares error $MSE = \frac{SSE}{N-k}$

Then, test statistic ($F$-statistic)  is:
$$ F = \frac{MSA}{MSE}  $$
